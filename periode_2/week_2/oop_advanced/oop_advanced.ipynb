## Polymorphism

Polymorphism simplifies coding by letting different types of objects share a common interface. It enables the use of a single method or function for various object types, enhancing code flexibility and reusability in object-oriented programming.

We can use the concept of polymorphism while creating class methods, as Python allows different classes to have methods with the same name. We can then later generalize calling these methods by disregarding the object we are working with, since different objects respond to the same method call in different ways (based on their specific implementations). Polymorphism promotes code flexibility, reusability, and extensibility.

For this upcoming section, let's redefine the linear function as $f$, where $f(x) = a \cdot x + b$, and the second-degree polynomial function as $g$, where $g(x) = a \cdot x^2 + b \cdot x + c$.

Now, suppose we want to create a new function, denoted as $h$, that combines a series of linear functions, followed by a series of polynomial functions. The length of both these series is variable. To illustrate, let's consider the following function, where three linear functions are followed by two polynomial functions:

$$h(x) = g_2(g_1(f_3(f_2(f_1(x)))))$$

Alternatively, we can express this as:

\begin{align}
f_1(x) &= y_1 \\
f_2(y_1) &= y_2 \\
f_3(y_2) &= y_3 \\
g_1(y_3) &= y_4 \\
g_2(y_4) &= y_5 = h(x)
\end{align}

Where the output of the innermost function ($f_1()$) is used as an input for the next function ($f_2()$), continuing until we are out of functions.

The overall process for function $h$ _always_ remains the same: linear functions are applied first, followed by polynomial functions. However, the number of linear or polynomial functions can vary.

In the cell below we have defined a `ChainLinearPolynomial` class, which we can use to simulate this process. 

````python
class ChainLinearPolynomial:
    def __init__(self, a, b, c, n_linear, n_polynomial):
        self.functions = []
        
        # append the same linear function n times
        linear = Linear(a, b)
        for _ in range(n_linear):
            self.functions.append(linear)

        # then append the same polynomial function n times
        polynomial = Polynomial(a, b, c)
        for _ in range(n_polynomial):
            self.functions.append(polynomial)

# following the example above, where we want the result of g_2(g_1(f_3(f_2(f_1(x)))))
chain = ChainLinearPolynomial(2, 3, 4, 3, 2) # a=2, b=3, c=4. There are 3 linear functions, then 2 polynomials

# set x to some value
x = 1

f_1 = chain.functions[0] # this is a function that defines: 2 * x + 3
f_1_y = f_1.calculate(x) # 2 * 1 + 3
print(f'f_1(x) \t\t\t\t\t= {f_1_y}') # which should give us 5

# f_1_y = f_1(x)
f_2 = chain.functions[1] # this is a function that defines: 2 * f_1_y + 3
f_2_y = f_2.calculate(f_1_y) # 2 * 5 + 3
print(f'f_2(f_1(x)) \t\t\t\t= {f_2_y}') # which should give us 13

# f_2_y = f_2(f_1(x))
f_3 = chain.functions[2] # this is a function that defines: 2 * f_2_y + 3
f_3_y = f_3.calculate(f_2_y) # 2 * 13 + 3
print(f'f_3(f_2(f_1(x))) \t\t\t= {f_3_y}') # which should give us 29

# f_3_y = f_3(f_2(f_1(x)))
g_1 = chain.functions[3] # this is a function that defines: 2 * f_3_y ** 2 + 3 * f_3_y + 4
g_1_y = g_1.calculate(f_3_y) # 2 * (29 ** 2) + 3 * 29 + 4
print(f'g_1(f_3(f_2(f_1(x)))) \t\t\t= {g_1_y}') # which should give us 1773

# g_1_y = g_1(f_3(f_2(f_1(x))))
g_2 = chain.functions[3] # this is a function that defines: 2 * g_1_y ** 2 + 3 * g_1_y + 4
h_x = g_2.calculate(g_1_y) # 2 * (? ** 2) + 3 * ? + 4
print(f'h(x) =  g_2(g_1(f_3(f_2(f_1(x))))) \t= {h_x}') # 6292381
````

## Inheritance

Inheritance is a concept in OOP that allows a class to inherit (copy) attributes and methods from another class. Inheritance is achieved by creating a new class (derived class, child class, or subclass) from an existing class (base class, parent class, or superclass). It provides a powerful mechanism for code reuse, promoting modularity, and creating relationships between classes. It allows you to build upon existing classes, and organize code in a hierarchical manner.

### Example

Let's say we aim to implement a logistic function, more specifically the sigmoid, defined as:
$$\sigma(a\cdot x + b) = \frac{1}{1 + e^{-(a\cdot x + b)}}$$

This function is essentially an extension of the linear function we previously created, with an additional calculation.

To avoid duplicating the properties of the linear class in this new class, we can use inheritance by creating a class that inherits from the `Linear` class. This approach allows us to make necessary adjustments to incorporate the logistic function functionality while building upon the existing features of the `Linear` class.

````python
from math import e

class LogisticLinear(Linear): # we inherit all methods Linear defined, so we don't need an __init__ here
    def calculate(self, x): # we want to override this method
        # first calculate the linear output
        y = super().calculate(x) # y = a * x + b

        # only need to define this part
        return 1 / (1 + e**(-y))
    
linear = Linear(1, 2)
logistic = LogisticLinear(1, 2)

linear.calculate(5), logistic.calculate(5)
````

To show that the `LogisticLinear` class did actually inherit from `Linear`, we can check whether `logistic` is an instance of both classes.

````python
isinstance(logistic, Linear), isinstance(logistic, LogisticLinear)
````
